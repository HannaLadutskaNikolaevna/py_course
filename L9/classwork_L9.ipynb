```python
class OfflineWallet:
    # init - Это метод конструктор
    def __init__(self, wid, owner, currency, volume):
        self.wid = wid
        self.owner = owner
        self.currency = currency
        self.volume = volume

    # метод экземпляра который меняет значение чегото
    def set_currency(self, new_currency):
        self.currency = new_currency
        print("Валюта счета изменена.")
    
    def __str__(self): # Для print
        return f"owner: {self.owner}\n\twid: {self.wid}, currency: {self.currency}, volume: {self.volume}"

    def __repr__(self): # для интерактивного режима >>>
        return f"repr\n\nOwner: {self.owner}\n\twid: {self.wid}, currency: {self.currency}, volume: {self.volume}"


class OnlineWallet(OfflineWallet):
    def __init__(self, wid, username, secret, owner, currency, volume):
        self.username = username
        self.secret = secret
        super().__init__(wid, owner, currency, volume)
    
    def online_pay(self, username, secret, volume):
        if self.username == username and self.secret == secret:
            self.volume -= volume
            print("Платеж прошел успешно")
        else:
             print("Платеж НЕ прошел успешно")
```
