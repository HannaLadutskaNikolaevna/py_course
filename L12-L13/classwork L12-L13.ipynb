```python
class User:
    def __init__(self, name):
        # self ==== переменная экземпляра
        self._name = name

    @property
    def name(self):
        print("get name")
        return self._name
    
    @name.setter
    def name(self, uname):
        print("set name")
        if isinstance(uname, str):
            self._name = uname
        print("new name is - ", self._name)

u12 = User("Dima")
print(u12.name)
u12.name = "sdasda"
```

```python
def my_print(*args, my_sep=" ", my_end="\n"):
    for value in args:
        print(value, end=my_sep)
    print(end=my_end)
    
```

```python
class User:
    def __init__(self, name):
        self.name = name

    def __iter__(self):
        return iter(self.name)

u12 = User("Dima")
for i in u12:
    print(i)
    
```

```python
class Student:
    def __init__(self, name, grades):
        self.name = name
        self.grades = grades
    def __iter__(self):
        print("Оценки студента - ", self.name)
        return iter(self.grades)

Dima = Student("Dima", [6, 7, 9, 4, 10])
for i in Dima:
    print(i)
    
```

```python
class Factorial:
    def __init__(self, max_number):
        self.max_number = max_number       

    def __iter__(self):
        self.start = 0
        return self

    def __next__(self):
        if self.start <= self.max_number:
            self.result = 1     
            
            if self.start == 0 or self.start == 1:
                self.start += 1
                return 1
                
            if self.start == 2:
                self.start += 1
                return 2

            for number in range(1, self.start+1):      
                self.result *= number

            self.start += 1
            return self.result
            
        raise StopIteration


for i in Factorial(5):
    print(i)
    
```

```python
def n(x):
    for i in range(x):
        yield i**2/2*4

for i in n(15):
    print(i)
    
```
